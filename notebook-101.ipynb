```python
df = spark.read.format("csv").option("header",True).load("/databricks-datasets/airlines/part-00000")
df.show(10)
```

```python
df.filter("Cancelled == 1").show(10)
```

```python
df_cancelled = df.filter("Cancelled == 1")
df_cancelled.write.mode("overwrite").saveAsTable("flight_delays_cancelled")
```

```python
%sql
select count(*) from flight_delays_cancelled;
```

```python
%sql
select dest,count(*) from flight_delays_cancelled group by Dest order by count(*) desc limit 10
```
